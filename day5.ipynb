```python
""" Q1 pattern 
1
12
123
1234
12345"""
for i in range(1,6,1):
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
"""Q2
12345
1234
123
12
1 """
for i in range(5,0,-1):
    for j in range(1,i+1):
        print(j,end=" ")
    print()    
```

```python
"""Q3
*****
****
***
**
* """
for i in range(5,0,-1):
    print("*"*i)
```

```python
"""Q4.
A
AB
ABC
ABCD
ABCDE"""
for i in range(1,6):
    for j in range(1,i+1):
        print(chr(64+j),end="")
    print()    
```
