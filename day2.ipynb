```python
#Q4.Check whether a year is a leap year
leap=int(input("enter a year : "))
if(leap%400==0):
    print("the year is leap year")
elif(leap%100==0):
    print("the year is not leap year")
elif(leap%4==0):
    print("the year is leap year")
else:
    print("not leap year")
```
