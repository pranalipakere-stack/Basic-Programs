```python
#area of square 
numb=int(input("enter a length"))
area= numb*numb
print("area of a square",area)
```

```python
#area of a rectangle 
l=int(input("enter a length"))
b=int(input("enter a breadth"))
area2=l*b
print("area of a rectangle",area2)
```

```python
#area of a triangle 
b2=int(input("enter a breadth"))
h=int(input("enter a hight"))
triangle=(1/2)*b2*h
print("area of a  triangle",triangle)
```

```python
#area of a circle 
r = float(input("enter a radius"))
circle=3.14*r*r            
print("area of a circle",circle) 
```

```python
#area of a Parallelogram
base=int(input("enter the base"))
height=int(input("enter the height"))
parallelogram=base*height
print("area of a parallelogram",parallelogram)
```

```python
#area of a Trapezium
a=int(input("enter a side1"))
b=int(input("enter a side2"))
h=int(input("enter a height"))
trapezium=(1/2)*(a+b)*h
print("area of Trapezium",trapezium)
```

```python
#area of Rhombus
d1=int(input("enter d1"))
d2=int(input("enter d2"))
Rhombus=(1/2)*d1*d2
print("area of a Rhombus",Rhombus)
```

```python
#area of Ellipse
a=int(input("enter a"))
b=int(input("enter b"))
Ellips= 3.14*a*b
print("area of ellips",Ellips)
```

```python
#area of cube
edge=int(input("enter a edge"))
cube=6*edge*edge
print("area of cube",cube)
```

```python
#area of cylinder 
r=float(input("enter the value of r"))
h=int(input("enter the value of h"))
cylinder=2*3.14*r*(r+h)
print("area of cylinder",cylinder)
```
